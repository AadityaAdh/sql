If you insert the first row with an explicit `id` value (e.g., `5000`) and then insert the second row without specifying an `id`, **MySQL will continue the auto-increment from the highest manually inserted `id`**.

### Example:
```sql
CREATE TABLE test_table (
    id INT AUTO_INCREMENT PRIMARY KEY,
    name VARCHAR(50)
);

-- Insert first row with a manually given id
INSERT INTO test_table (id, name) VALUES (5000, 'Alice');

-- Insert second row without specifying id
INSERT INTO test_table (name) VALUES ('Bob');
```

### What happens?
- The first row gets `id = 5000`.
- The second row, which does not specify an `id`, will get `id = 5001` (because MySQL sets the **next auto-increment value to `MAX(id) + 1`**).
  
So, the result will be:
```
id   | name
----------------
5000 | Alice
5001 | Bob
```

### Key Behavior:
- If the table was **empty** before inserting `5000`, MySQL would normally start from `1`, but since we manually inserted `5000`, the next auto-generated `id` will be `5001`.
- If you manually insert a lower value (e.g., `1000`), MySQL **does not** reset the auto-increment to `1001`. It keeps track of the highest value.
