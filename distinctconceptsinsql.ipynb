The **syntax of `DISTINCT`** in SQL is straightforward. It is used to eliminate duplicate rows from the result set, returning only unique values.

### General Syntax of `DISTINCT`:

```sql
SELECT DISTINCT column1, column2, ...
FROM table_name;
```

### Explanation:
- **`DISTINCT`**: Specifies that only unique values should be returned.
- **`column1, column2, ...`**: A comma-separated list of columns you want to retrieve distinct values for.
- **`table_name`**: The name of the table you are selecting data from.

### Example with One Column:
```sql
SELECT DISTINCT name
FROM employee;
```
- This will return a list of **unique names** from the `employee` table, eliminating duplicates.

### Example with Multiple Columns:
```sql
SELECT DISTINCT name, departmentid
FROM employee;
```
- This will return **unique pairs** of `(name, departmentid)` combinations. If `name = 'John'` and `departmentid = 'Sales'` appears multiple times, it will only appear once in the result.

### Example with `JOIN`:
```sql
SELECT DISTINCT e.name, t.name
FROM employee AS e
JOIN employee AS t
    ON e.salary = t.salary
    AND e.departmentid = t.departmentid
WHERE e.name <> t.name;
```
- This will return **unique pairs** of names from two instances of the `employee` table, where `e.name` is different from `t.name` and both have the same salary and department.

### Key Points:
1. **`DISTINCT` applies to the entire result row**: For example, `DISTINCT column1, column2` will return unique pairs (tuples) based on the combination of both columns.
2. If there are duplicates for the combination of columns, they will be eliminated from the result set.

### To clarify:
- **`DISTINCT`** eliminates duplicate rows from the result, but it works across **all the columns** you specify in the `SELECT` clause.
- If you specify multiple columns, **the combination of values from all those columns is treated as a unit**, and SQL will return only the unique combinations.

### Example with Multiple Columns:
Let’s say you have the following data in the `employee` table:

| name   | departmentid | salary |
|--------|--------------|--------|

Now, if you want to select **distinct pairs** of `name` and `departmentid`, the query would be:

```sql
SELECT DISTINCT name, departmentid
FROM employee;
```

### Result:
| name   | departmentid |

| Alice  | HR           |
| Bob    | IT           |
| John   | HR           |

- **Explanation**: Even though there were duplicates (like `Alice` with `HR` and `Bob` with `IT`), SQL will return only the **unique combinations** of `name` and `departmentid`. So, only one row for each unique combination appears in the result.

### If You Add More Columns:
If you add another column, say `salary`, the distinct combinations will include all three columns:

```sql
SELECT DISTINCT name, departmentid, salary
FROM employee;
```

### Result:
| name   | departmentid | salary |
|--------|--------------|--------|
| Alice  | HR           | 5000   |
| Bob    | IT           | 6000   |
| John   | HR           | 7000   |

- **Explanation**: Now, SQL is ensuring the **distinctness** of combinations across all three columns: `name`, `departmentid`, and `salary`. So, even if the same `name` and `departmentid` combination exists with different salaries, it will be considered unique.
